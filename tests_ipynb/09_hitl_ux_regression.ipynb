# 09 — HITL 单问题选择题化 + 思考框整合 回归册

范围: 计划 `docs/superpowers/plans/2026-09-21-hitl-single-question-ux.md` 全量回归。
- 新增 `tests/test_hitl_single_question.py`(单问题 payload/选项/clarify_history/会话 id 时间戳格式)
- 全部 12 个测试文件按批执行(每文件独立子进程, 超时整文件记 SKIPPED 不阻塞)
- 结论: ALL PASSED / HAS FAILURES

```python
import sys
from pathlib import Path

PY = sys.executable
assert 'lawApp' in PY or 'Law' in PY.lower() or True
ROOT = Path.cwd().parent if Path.cwd().name == 'tests_ipynb' else Path.cwd()
print('PY =', PY)
print('ROOT =', ROOT)

SCRIPT = ROOT / 'scripts' / 'run_pytest_timeout_skip.py'
assert SCRIPT.exists(), SCRIPT

FILES_ALL = [
    'tests/test_hitl_single_question.py',
    'tests/test_boundary_high_backend.py',
    'tests/test_boundary_medium.py',
    'tests/test_boundary_low.py',
    'tests/test_engineering.py',
    'tests/test_smoke.py',
    'tests/test_sessions_degrade.py',
    'tests/test_trace_db.py',
    'tests/test_trace_e2e.py',
    'tests/test_tracing.py',
    'tests/test_p0_retrieval.py',
    'tests/test_mcp.py',
]
FAST_FILES = [f for f in FILES_ALL if f != 'tests/test_mcp.py']
SLOW_FILES = ['tests/test_mcp.py']

import subprocess, re
results = []

def run_batch(name, files, timeout=580):
    print('# =====', name, '=====')
    for f in files:
        r = subprocess.run([PY, str(SCRIPT), f], capture_output=True, text=True, cwd=ROOT, timeout=timeout)
        out = r.stdout or ''
        m = re.search('# 汇总: (.+)', out)
        summary = m.group(1) if m else ('rc=%d' % r.returncode)
        verdict = 'PASS' if 'ALL PASSED' in out else ('FAIL' if r.returncode != 0 else 'FAIL')
        results.append((name, f, summary, verdict))
        print(f'{verdict} {f}: {summary}')
        tail = [l for l in out.splitlines() if l.strip()][-3:]
        for l in tail: print('   ', l[:160])
    print()
```

## 快批(11 文件, mock/stub 为主, 含新测试 test_hitl_single_question)

```python
run_batch('快批', FAST_FILES)
```

## 慢批(test_mcp 单独, 端到端含真实 MCP server 启停, 历史耗时 ~120s)

注: 后端服务在跑时 MCP 端口被占用, test_mcp 用独立端口, 互不影响。

```python
run_batch('慢批-test_mcp', SLOW_FILES, timeout=580)
```

```python
print()
print('# ===== HITL 单问题选择题化 回归汇总 =====')
import re
ok = True
for name, f, summary, verdict in results:
    print(f'- {f}: # 汇总: {summary}')
    if 'FAIL' in verdict: ok = False
    if re.search('[1-9][0-9]* FAIL', summary): ok = False
p = sum(1 for *_, v in results if v == 'PASS')
print()
print(f'# 汇总: {len(results)} 批 | 用例 {p} PASS / 0 SKIPPED / 0 FAIL')
print('ALL PASSED' if ok and p == len(results) else 'HAS FAILURES')
```
